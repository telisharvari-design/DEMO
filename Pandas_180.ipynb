**Practice Scenario: Consider New York state and check for startups whose investment in MKT is less 150000**

```python
fil4 = df['STATE'] =='New York'
fil5 = df['MKT']<150000
df[fil4 & fil5]
```

**Consider startup from either New York or California whose investment in ADMIN is more than 100000**

combination of OR and AND logical conditions

OR is between New York and california

AND is between the above result and ADMIN>100000

```python
fil6 = (df['STATE']=='California') | (df['STATE']=='New York')
fil7 = df['ADMIN']>100000
df[fil6 & fil7]
```

**Practice scenario: Consider startups from Florida state whose profit is ranging between 125000 and 180000**

```python
fil8 = df['STATE']== 'Florida'
fil9 = df['PROFIT'].between(125000,180000)
df[fil8 & fil9]
```

**Sort**

syntax: asc: df.sort_values(by=['column_name'])

Ascending or Descending order

1) Numerical data: smallest to largest or vice versa
2) Text data: ascending: a-z or descending: z-a

**agg({column_name:aggfunc})**

Display the minimum,maximum,sum and average profit in the original dataframe

```python
df.agg({"PROFIT":['min','max','sum','mean']})
```

**Display the min,max profits made by startups that belong to Florida state**

```python
# first i have to filter my state and then we can fetch the min max details
df.query("STATE=='Florida'").agg({'PROFIT':['min','max']})
```

```python
df.query("STATE=='Florida'").sort_values(by='PROFIT').head(1)[['PROFIT']]
```

```python
df.query("STATE=='Florida'").sort_values(by='PROFIT').tail(1)[['PROFIT']]
```

```python
df.query("STATE=='Florida'").agg({'PROFIT':['min','max','std'],
                                  'ADMIN':['min','max','std']})
```

**Practice Scenario: Display std and total profit of startups that belong to California state**

Group BY

```python
df.groupby(by=['STATE']).agg({'PROFIT':['min','max']})
```

```python
# first filter data related to state belonging to california and florida
df_2states = df.query("(STATE=='California') | (STATE=='Florida')")
df_2states.groupby(by='STATE').agg({'PROFIT':['sum'],
                                    'RND':['sum']})
```
