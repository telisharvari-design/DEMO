## Statistical and Aggregation Operations
    
    prod()
    sum()
    min()
    max()
    mean()
    std()
    median() : centre most point in the data

```python
import numpy as np

# create an array of profit of the products
profit = np.array([45000,3200,7600,50000,3600,12000])
print(profit)

# can you calculate the cumulative profit made by all the products
culumative_profit = np.sum(profit) # 45000 + 3200+...
print(culumative_profit)

# can you calculate the least profit
least = np.min(profit)
print(least)

# can you calculate the maximum profit received
max_profit = np.max(profit)
print(max_profit)

# can you calculate average profit
avg_profit = np.mean(profit)
print(avg_profit)
print(round(avg_profit,2))
```

```python
# can you calculate the centre most profit area which is observed from the current dataset
print(np.median(profit))
```

```python
## standard Deviation : gives you an idea by what value
## is mean differing from each datapoint
## average - datapoint
```

```python
print(np.mean(profit))
print(np.std(profit))
print(np.var(profit))
```

```python
# factorial : multiple all the elements: prod()
data_elements = np.array([1,2,10])
print(np.prod(data_elements))
# 1*2*10
```

```python
# 25% percentile of data
print(np.percentile(profit,0.25))
```
